```python
# ============================================================
# ATIVIDADE 4 - Modelagem de um problema real
# Tema escolhido: motorista de aplicativo escolhendo quais corridas aceitar
# Objetivo: aplicar os conceitos de espaco de busca, funcao objetivo
# e restricoes vistos na aula em um problema do dia a dia
# ============================================================
import random

# ----------------------------------------------------------
# 1. DEFINICAO DO PROBLEMA
# ----------------------------------------------------------
# No comeco do turno, o motorista recebe uma lista de corridas
# disponiveis. Cada corrida tem uma duracao (em minutos) e um
# ganho (em reais). O motorista quer decidir quais corridas aceitar
# para ganhar o maximo possivel sem ultrapassar o tempo que ele
# tem disponivel no turno.

random.seed(7)

n_corridas = 10
duracao = [random.randint(10, 40) for _ in range(n_corridas)]          # minutos
ganho = [round(random.uniform(8, 35), 2) for _ in range(n_corridas)]   # reais
tempo_turno = 180  # minutos disponiveis no turno (3 horas)

# ----------------------------------------------------------
# 2. REPRESENTACAO DE UMA SOLUCAO
# ----------------------------------------------------------
# Uma solucao candidata e um vetor de 0s e 1s do tamanho n_corridas.
# 1 = motorista aceita aquela corrida, 0 = motorista recusa.
# Exemplo com 10 corridas: [1,0,0,1,1,0,1,0,0,1]

def gerar_solucao_aleatoria(n):
    return [random.randint(0, 1) for _ in range(n)]

# ----------------------------------------------------------
# 3. FUNCAO OBJETIVO
# ----------------------------------------------------------
# Soma o ganho de todas as corridas aceitas. Isso e o que o
# motorista quer maximizar.

def funcao_objetivo(solucao, ganho):
    return sum(g for s, g in zip(solucao, ganho) if s == 1)

# ----------------------------------------------------------
# 4. VERIFICACAO DAS RESTRICOES
# ----------------------------------------------------------
# A soma da duracao das corridas aceitas nao pode ultrapassar o
# tempo disponivel no turno.

def eh_factivel(solucao, duracao, tempo_turno):
    tempo_usado = sum(d for s, d in zip(solucao, duracao) if s == 1)
    return (tempo_usado <= tempo_turno), tempo_usado

# ----------------------------------------------------------
# 5. TESTANDO COM UMA SOLUCAO ALEATORIA
# ----------------------------------------------------------
solucao = gerar_solucao_aleatoria(n_corridas)
ganho_total = funcao_objetivo(solucao, ganho)
factivel, tempo_usado = eh_factivel(solucao, duracao, tempo_turno)

print('Corridas disponiveis no turno:')
for i in range(n_corridas):
    print(f'  Corrida {i}: duracao={duracao[i]:2d} min | ganho=R${ganho[i]:.2f}')

print('\nSolucao aleatoria gerada (1=aceita, 0=recusa):', solucao)
print('Tempo total usado:', tempo_usado, 'min (limite do turno:', tempo_turno, 'min)')
print('Ganho total da solucao: R$', round(ganho_total, 2))
print('Essa solucao respeita a restricao de tempo?', factivel)

# ----------------------------------------------------------
# 6. GERANDO ALGUMAS SOLUCOES PARA COMPARAR
# ----------------------------------------------------------
print('\nGerando mais 4 solucoes aleatorias para comparacao:')
for tentativa in range(4):
    sol = gerar_solucao_aleatoria(n_corridas)
    g = funcao_objetivo(sol, ganho)
    ok, t_usado = eh_factivel(sol, duracao, tempo_turno)
    status = 'FACTIVEL' if ok else 'INVALIDA (estourou o tempo)'
    print(f'  Tentativa {tentativa+1}: ganho=R${g:6.2f} | tempo={t_usado:3d} min | {status}')
```
